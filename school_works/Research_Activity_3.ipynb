```python
print("Missing Code values:",
      df["Code"].isna().sum())

print("\nCode length counts:")
print(df["Code"].dropna().str.len().value_counts())

Q1 = df["Population (historical estimates)"].quantile(0.25)
Q3 = df["Population (historical estimates)"].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR

outliers = df[
    df["Population (historical estimates)"] > upper_bound
]

print("\nPotential population outliers:", len(outliers))

print("\nNegative population values:",
      (df["Population (historical estimates)"] < 0).sum())

print("Duplicate Entity-Year records:",
      df.duplicated(subset=["Entity", "Year"]).sum())
```

```python
cleaned_df = df.copy()

cleaned_df["Entity"] = cleaned_df["Entity"].str.strip()

cleaned_df["Code"] = cleaned_df["Code"].fillna("N/A")
cleaned_df["Code"] = cleaned_df["Code"].astype(str).str.strip()

cleaned_df = cleaned_df.drop_duplicates()

cleaned_df["Year"] = pd.to_numeric(
    cleaned_df["Year"],
    errors="coerce"
)

cleaned_df["Population (historical estimates)"] = pd.to_numeric(
    cleaned_df["Population (historical estimates)"],
    errors="coerce"
)

print("Cleaning completed!")
```

```python
print("Original shape:", raw_df.shape)
print("Cleaned shape:", cleaned_df.shape)

print("\nMissing values after cleaning:")
print(cleaned_df.isnull().sum())

print("\nDuplicate rows after cleaning:",
      cleaned_df.duplicated().sum())

print("\nFirst 5 cleaned records:")
display(cleaned_df.head())
```

```python
cleaned_df.to_csv("cleaned_data.csv", index=False)

print("cleaned_data.csv created successfully!")
```
